```python
# Tarea 3 - Las cámaras de frío, con variables

## 1. Variables

tiendas = 120
camaras_por_tiendas = 3
lecturas_por_hora = 12

## 2. Calcula las lecturas al dia
lecturas_al_dia = tiendas * camaras_por_tiendas * lecturas_por_hora * 24

## 3. Calculo de lecturas al año
lecturas_al_ano = lecturas_al_dia * 365

## 4. Mostrar los resultados con print y una f-string
print(f"lecturas al dia: {lecturas_al_dia}")
print(f"lecturas al ano: {lecturas_al_ano}")






```

```python
#Tarea 4 - De lecturas a megas

bytes_por_lectura = 46

# 2. Calcula megas al dia  (1 MB = 1000000)

megas_al_dia = (lecturas_al_dia* bytes_por_lectura) / (1000000)

# 3. Calcula gigas al anio ( 1GB = 1000 MB)

gigas_al_ano = (megas_al_dia * 365)/1000

# 4. muestra los resultados con dos decimales

print(f"sensores:{megas_al_dia:.2f} MB al dia")
print(f"sensores:{gigas_al_ano:.2f} GB al ano")



```

```python
# Tarea 5 - Los clics de la web y la app

visitas_al_dia = 250000
clics_por_visita = 12
bytes_por_clip = 220

# 1. Clics al dia

clic_al_dia = visitas_al_dia * clics_por_visita

# 2. Megas ocupados

megas_al_dia = (clic_al_dia * bytes_por_clip) / (1000000)

# 3. Gigas ocupadas al ano

gigas_al_ano = (megas_al_dia * 365) / 1000

# Resultados
print(f"clics al dia: {clic_al_dia:,}")
print(f"clics:{megas_al_dia:.0f} MB al dia")
print (f"clics:{gigas_al_ano:.1f} GB al ano")



```
